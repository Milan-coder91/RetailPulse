```python
# Day2 

# Duplicate rows remove karo
for name, df in datasets.items():
    before = len(df)
    datasets[name] = df.drop_duplicates().copy()
    after = len(datasets[name])

    print(f"{name}: {before - after} duplicate rows removed")

# Missing values report
for name, df in datasets.items():
    print(f"\n{name.upper()} - Missing Values")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
for name, df in datasets.items():
    for col in df.columns:
        if df[col].isnull().any():
            # ID columns ko automatically fill nahi karna
            if "id" in col.lower():
                continue

            if pd.api.types.is_numeric_dtype(df[col]):
                median_value = df[col].median()
                if pd.notna(median_value):
                    df[col] = df[col].fillna(median_value)

            else:
                mode_value = df[col].mode()
                if not mode_value.empty:
                    df[col] = df[col].fillna(mode_value[0])

    datasets[name] = df

print("Missing values handled where appropriate!")
```

```python
# Sales dataset ko columns se identify karo
sales_candidates = [
    df for df in datasets.values()
    if {"date", "receipt_id", "customer_id", "total_value"}.issubset(df.columns)
]

if not sales_candidates:
    raise ValueError("Sales dataset nahi mila. Dataset columns check karo.")

sales = sales_candidates[0].copy()

sales["date"] = pd.to_datetime(sales["date"], errors="coerce")

print(sales.head())
```

```python
# Invalid date ya customer ID wali rows hatao
sales_rfm = sales.dropna(
    subset=["date", "customer_id", "receipt_id", "total_value"]
).copy()

reference_date = sales_rfm["date"].max() + pd.Timedelta(days=1)

rfm = sales_rfm.groupby("customer_id").agg(
    Recency=("date", lambda x: (reference_date - x.max()).days),
    Frequency=("receipt_id", "nunique"),
    Monetary=("total_value", "sum")
).reset_index()

print(rfm.head())
```

```python
# 4 groups mein RFM scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4, labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4, labels=[1, 2, 3, 4]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4, labels=[1, 2, 3, 4]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

print(rfm.head())
```

```python
# Daily sales calculate karo
daily_sales = (
    sales.dropna(subset=["date", "total_value"])
    .groupby(pd.Grouper(key="date", freq="D"))["total_value"]
    .sum()
    .reset_index()
    .sort_values("date")
)

# 7-day aur 30-day rolling average
daily_sales["Rolling_7_Day"] = (
    daily_sales["total_value"].rolling(7, min_periods=1).mean()
)

daily_sales["Rolling_30_Day"] = (
    daily_sales["total_value"].rolling(30, min_periods=1).mean()
)

print(daily_sales.tail())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
```

```python
features = ["Recency", "Frequency", "Monetary"]

X = rfm[features].copy()

X = X.replace([np.inf, -np.inf], np.nan)
X = X.dropna()

rfm = rfm.loc[X.index].copy()

print("Features:", features)
print("Shape:", X.shape)
display(X.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaling completed!")
print("Scaled data shape:", X_scaled.shape)
```

```python
# K-Means clustering


kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["KMeans_Cluster"] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")
print(rfm["KMeans_Cluster"].value_counts().sort_index())
```

```python
# DBSCAN clustering

dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

rfm["DBSCAN_Cluster"] = dbscan.fit_predict(X_scaled)

print("DBSCAN clustering completed!")
print(rfm["DBSCAN_Cluster"].value_counts().sort_index())
```

```python
# K-Means Cluster Evaluation


kmeans_score = silhouette_score(
    X_scaled,
    rfm["KMeans_Cluster"]
)

print("K-Means Silhouette Score:", round(kmeans_score, 4))
```

```python
# DBSCAN Cluster Evaluation


labels = rfm["DBSCAN_Cluster"]
mask = labels != -1

if labels[mask].nunique() >= 2:
    dbscan_score = silhouette_score(
        X_scaled[mask],
        labels[mask]
    )
    print("DBSCAN Silhouette Score:", round(dbscan_score, 4))
else:
    print("Silhouette Score calculate nahi ho sakta: 2 clusters required.")
```

```python
# K-Means cluster summary

cluster_summary = rfm.groupby("KMeans_Cluster").agg(
    Customers=("customer_id", "count"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Monetary=("Monetary", "mean")
).round(2)

display(cluster_summary)
```

```python
# Business Interpretation


def interpret_cluster(row):
    if (
        row["Avg_Recency"] <= rfm["Recency"].median()
        and row["Avg_Frequency"] >= rfm["Frequency"].median()
        and row["Avg_Monetary"] >= rfm["Monetary"].median()
    ):
        return "Loyal / High-Value Customers"

    elif row["Avg_Recency"] > rfm["Recency"].median():
        return "At-Risk / Inactive Customers"

    elif row["Avg_Frequency"] <= rfm["Frequency"].median():
        return "Occasional Customers"

    else:
        return "Regular Customers"


cluster_summary["Business_Segment"] = cluster_summary.apply(
    interpret_cluster, axis=1
)

display(cluster_summary)
```

```python
split_index = int(len(daily_demand) * 0.8)

train_data = daily_demand.iloc[:split_index].copy()
test_data = daily_demand.iloc[split_index:].copy()

print("Training records:", len(train_data))
print("Testing records:", len(test_data))

display(train_data.tail())
display(test_data.head())
```

```python
model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=True,
    daily_seasonality=False
)

model.fit(train_data)

print("Prophet baseline model training completed!")
```

```python
future_test = test_data[["ds"]].copy()

forecast_test = model.predict(future_test)

test_results = test_data.merge(
    forecast_test[["ds", "yhat", "yhat_lower", "yhat_upper"]],
    on="ds",
    how="left"
)

display(test_results.head())
```

```python
WINDOW_SIZE = 14

def create_sequences(data, window=14):
    X, y = [], []

    for i in range(window, len(data)):
        X.append(data[i-window:i])
        y.append(data[i])

    return (
        np.array(X, dtype=np.float32),
        np.array(y, dtype=np.float32)
    )

X_train, y_train = create_sequences(train_values, WINDOW_SIZE)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```python
class DemandDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, index):
        return self.X[index], self.y[index]


train_dataset = DemandDataset(X_train, y_train)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

print("Training DataLoader ready!")
```

```python
class LSTMModel(pl.LightningModule):
    def __init__(self, hidden_size=64, learning_rate=0.001):
        super().__init__()

        self.save_hyperparameters()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        self.fc = nn.Linear(hidden_size, 1)
        self.loss_fn = nn.MSELoss()

    def forward(self, x):
        output, _ = self.lstm(x)
        return self.fc(output[:, -1, :])

    def training_step(self, batch, batch_idx):
        x, y = batch
        prediction = self(x)
        loss = self.loss_fn(prediction, y)
        self.log("train_loss", loss, prog_bar=True)
        return loss

    def configure_optimizers(self):
        return torch.optim.Adam(
            self.parameters(),
            lr=self.hparams.learning_rate
        )


model = LSTMModel()

print(model)
```

```python
trainer = pl.Trainer(
    max_epochs=10,
    accelerator="auto",
    devices=1,
    enable_checkpointing=False,
    logger=False,
    enable_progress_bar=True
)

trainer.fit(model, train_loader)

print("LSTM training completed!")
```

```python
test_scaled = all_scaled[split_index - WINDOW_SIZE:]

X_test, y_test = create_sequences(
    test_scaled,
    WINDOW_SIZE
)

model.eval()

with torch.no_grad():
    X_tensor = torch.tensor(X_test, dtype=torch.float32)
    predictions_scaled = model(X_tensor).cpu().numpy()

predictions = scaler.inverse_transform(predictions_scaled)
actual = scaler.inverse_transform(y_test)

print("Prediction completed!")
print("Test predictions:", len(predictions))
```

```python
mae = mean_absolute_error(actual, predictions)

rmse = np.sqrt(
    mean_squared_error(actual, predictions)
)

print("LSTM Model Evaluation")
print("---------------------")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
# Features Ebgineering 


CUTOFF_DATE = pd.Timestamp("2025-09-30")
FUTURE_END_DATE = CUTOFF_DATE + pd.Timedelta(days=90)

print("Feature cutoff:", CUTOFF_DATE.date())
print("Churn observation period:", 
      (CUTOFF_DATE + pd.Timedelta(days=1)).date(),
      "to",
      FUTURE_END_DATE.date())
```

```python
# Sales before cutoff
history = sales[
    sales["date"] <= CUTOFF_DATE
].copy()

# Future 90-day transactions
future = sales[
    (sales["date"] > CUTOFF_DATE) &
    (sales["date"] <= FUTURE_END_DATE)
].copy()

print("Historical transactions:", len(history))
print("Future transactions:", len(future))
```

```python
# Customer Lavel Features 

customer_features = (
    history
    .groupby("customer_id")
    .agg(
        last_purchase_date=("date", "max"),
        frequency=("receipt_id", "nunique"),
        monetary=("total_value", "sum"),
        total_quantity=("quantity", "sum"),
        avg_order_value=("total_value", "mean"),
        avg_quantity=("quantity", "mean"),
        unique_skus=("sku_id", "nunique"),
        unique_stores=("store_id", "nunique"),
        active_months=("date", lambda x: x.dt.to_period("M").nunique())
    )
    .reset_index()
)

# Recency
customer_features["recency"] = (
    CUTOFF_DATE -
    customer_features["last_purchase_date"]
).dt.days

print("Customer feature shape:", customer_features.shape)

display(customer_features.head())
```

```python
#Promotion usage feature

promo_features = (
    history
    .assign(
        promo_used=history["promo_id"].notna().astype(int)
    )
    .groupby("customer_id")
    .agg(
        promo_transactions=("promo_used", "sum"),
        promo_usage_rate=("promo_used", "mean")
    )
    .reset_index()
)

customer_features = customer_features.merge(
    promo_features,
    on="customer_id",
    how="left"
)

customer_features[
    ["promo_transactions", "promo_usage_rate"]
] = customer_features[
    ["promo_transactions", "promo_usage_rate"]
].fillna(0)

display(customer_features.head())
```

```python
# Create churn Target 

future_customers = set(
    future["customer_id"].dropna().unique()
)

customer_features["churn"] = (
    ~customer_features["customer_id"]
    .isin(future_customers)
).astype(int)

print("Churn distribution:")
print(
    customer_features["churn"]
    .value_counts()
)

print("\nChurn percentage:")
print(
    customer_features["churn"]
    .mean() * 100
)
```

```python

# DAY 11 - STEP 12
# Build existing customer features for the 2,000 labeled customers

sales = pd.read_csv(BASE_PATH / "cleaned_sales_transactions.csv")
sales["date"] = pd.to_datetime(sales["date"], errors="coerce")

labeled_ids = set(churn_predictions["customer_id"])

sales = sales[sales["customer_id"].isin(labeled_ids)].copy()

reference_date = sales["date"].max() + pd.Timedelta(days=1)

customer_features = sales.groupby("customer_id").agg(
    frequency=("receipt_id", "nunique"),
    monetary=("total_value", "sum"),
    total_quantity=("quantity", "sum"),
    avg_order_value=("total_value", "mean"),
    avg_quantity=("quantity", "mean"),
    unique_skus=("sku_id", "nunique"),
    unique_stores=("store_id", "nunique"),
    active_months=("date", lambda x: x.dt.to_period("M").nunique()),
    recency=("date", lambda x: (reference_date - x.max()).days),
    promo_transactions=("promo_id", lambda x: x.notna().sum())
).reset_index()

customer_features["promo_usage_rate"] = (
    customer_features["promo_transactions"] /
    customer_features["frequency"].replace(0, np.nan)
).fillna(0)

tuning_data = customer_features.merge(
    churn_predictions[["customer_id", "actual_churn"]],
    on="customer_id",
    how="inner"
)

feature_cols = model.get_booster().feature_names

print("Tuning dataset shape:", tuning_data.shape)
print("\nFeature columns available:", 
      all(col in tuning_data.columns for col in feature_cols))
print("Missing values in model features:",
      int(tuning_data[feature_cols].isna().sum().sum()))
print("\nChurn label counts:")
print(tuning_data["actual_churn"].value_counts())

```

```python

# DAY 11 - STEP 13
# Optuna hyperparameter tuning on the prepared dataset
# The saved Day 9 model will not be overwritten

import optuna
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from xgboost import XGBClassifier

X = tuning_data[feature_cols]
y = tuning_data["actual_churn"]

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

def objective(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 400),
        "max_depth": trial.suggest_int("max_depth", 2, 8),
        "learning_rate": trial.suggest_float(
            "learning_rate", 0.01, 0.2, log=True
        ),
        "subsample": trial.suggest_float("subsample", 0.7, 1.0),
        "colsample_bytree": trial.suggest_float(
            "colsample_bytree", 0.7, 1.0
        ),
        "scale_pos_weight": trial.suggest_float(
            "scale_pos_weight", 5.0, 30.0
        ),
        "eval_metric": "logloss",
        "random_state": 42,
        "n_jobs": -1
    }

    candidate = XGBClassifier(**params)
    candidate.fit(X_train, y_train)

    probabilities = candidate.predict_proba(X_valid)[:, 1]
    predictions = (probabilities >= 0.5).astype(int)

    return f1_score(y_valid, predictions, zero_division=0)

study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=20)

print("Best validation F1-score:", round(study.best_value, 4))
print("\nBest hyperparameters:")
print(study.best_params)
print("\nTrials completed:", len(study.trials))

```

```python

# DAY 11 - STEP 14
# Train final model using best Optuna parameters and save separately

import json

best_params = study.best_params.copy()

tuned_model = XGBClassifier(
    **best_params,
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

tuned_model.fit(X_train, y_train)

model_output = BASE_PATH / "day11_tuned_churn_model.json"
tuned_model.save_model(str(model_output))

params_output = BASE_PATH / "day11_optuna_best_params.json"
with open(params_output, "w", encoding="utf-8") as f:
    json.dump(best_params, f, indent=4)

print("Tuned model saved:", model_output.name)
print("Best parameters saved:", params_output.name)
print("Validation F1-score:", round(study.best_value, 4))
print("Original Day 9 model remains unchanged.")

```

```python
# Day 14 — Step 31: Restore LSTM Model Class

import torch
import torch.nn as nn
import pytorch_lightning as pl

class LSTMModel(pl.LightningModule):
    def __init__(self, hidden_size=64, learning_rate=0.001):
        super().__init__()
        self.save_hyperparameters()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        self.fc = nn.Linear(hidden_size, 1)
        self.loss_fn = nn.MSELoss()

    def forward(self, x):
        output, _ = self.lstm(x)
        return self.fc(output[:, -1, :])

    def training_step(self, batch, batch_idx):
        x, y = batch
        prediction = self(x)
        loss = self.loss_fn(prediction, y)
        self.log("train_loss", loss, prog_bar=True)
        return loss

    def configure_optimizers(self):
        return torch.optim.Adam(
            self.parameters(),
            lr=self.hparams.learning_rate
        )

print("LSTMModel class is ready.")
```

```python
# Day 14 — Step 32: Rebuild LSTM Training Loader

import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader

def create_sequences(data, window=14):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i])
        y.append(data[i])
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.float32)

class DemandDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, index):
        return self.X[index], self.y[index]

X_train, y_train = create_sequences(train_values, WINDOW_SIZE)

train_dataset = DemandDataset(X_train, y_train)
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("Training batches:", len(train_loader))
print("Training loader ready.")
```

```python
# Day 14 — Step 42: Prepare Correct LSTM Training Data

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

correct_sales = pd.read_csv(
    BASE_PATH / "day14_correct_daily_sales.csv"
)

correct_sales["date"] = pd.to_datetime(correct_sales["date"])
correct_sales["total_value"] = pd.to_numeric(
    correct_sales["total_value"], errors="coerce"
)

daily_demand = (
    correct_sales.dropna(subset=["date", "total_value"])
    .groupby("date")["total_value"]
    .sum()
    .sort_index()
)

values = daily_demand.values.reshape(-1, 1)
WINDOW_SIZE = 14
split_index = int(len(values) * 0.8)

scaler = MinMaxScaler()
train_values = scaler.fit_transform(values[:split_index])
all_scaled = scaler.transform(values)

print("Total days:", len(daily_demand))
print("Training days:", len(train_values))
print("Testing days:", len(values) - split_index)
print("Demand minimum:", values.min())
print("Demand maximum:", values.max())
print("Training data ready.")
```

```python
# Day 14 — Step 43: Build LSTM Training Sequences

import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader

def create_sequences(data, window=14):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i])
        y.append(data[i])
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.float32)

X_train, y_train = create_sequences(train_values, WINDOW_SIZE)

class DemandDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, index):
        return self.X[index], self.y[index]

train_dataset = DemandDataset(X_train, y_train)
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("Training batches:", len(train_loader))
print("Training loader ready.")
```

```python
# Day 15 — Step 17: Create Dashboard Dataset Summary

import pandas as pd

dataset_names = [
    "cleaned_sales_transactions.csv",
    "cleaned_customer_master.csv",
    "cleaned_inventory_snapshot.csv",
    "cleaned_promotions.csv",
    "cleaned_sku_master.csv",
    "cleaned_store_master.csv",
    "cleaned_sku_inventory_flags.csv"
]

summary = []

for name in dataset_names:
    file_path = BASE_PATH / name

    if file_path.exists():
        df = pd.read_csv(file_path, nrows=0)
        row_count = sum(1 for _ in open(file_path, encoding="utf-8")) - 1

        summary.append({
            "Dataset": name,
            "Rows": row_count,
            "Columns": len(df.columns),
            "Status": "Available"
        })

dataset_summary = pd.DataFrame(summary)
display(dataset_summary)
```

```python
# Step 18: Save Dashboard Dataset Summary

output_path = BASE_PATH / "dashboard_dataset_summary.csv"

dataset_summary.to_csv(output_path, index=False)

print("Dataset summary saved successfully!")
print("File location:", output_path)

```
